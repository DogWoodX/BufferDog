#### 4h

```python
# @title
import pandas as pd
import plotly.graph_objs as go

# Load the data
file_path = '/content/drive/MyDrive/Colab Notebooks/Data Sets/4hFTIR.csv'
df = pd.read_csv(file_path)

# Define the wavenumber column and the columns to plot
wavenumber = df['ABS']
columns_to_plot = df.columns[df.columns != 'ABS']  # Exclude the wavenumber column

# Find the global minimum value across all columns
global_min_value = df[columns_to_plot].min().min()

# Create a Plotly figure
fig = go.Figure()

# Add traces for each column, shifting each series so they all share the same baseline
for column in columns_to_plot:
    # Calculate the minimum value for the current column
    column_min_value = df[column].min()

    # Adjust each series by subtracting its minimum value and then adding the global minimum value
    y_values = df[column] - column_min_value + global_min_value

    fig.add_trace(go.Scatter(
        x=wavenumber,
        y=-y_values,  # Invert to maintain the downward orientation
        mode='lines',
        name=column
    ))

# Update layout to match the required style
fig.update_layout(
    yaxis=dict(
        title="Relative Signal",
        showgrid=True,
        showticklabels=False,  # Hide Y-axis numbers
    ),
    xaxis=dict(
        title="Wavenumber",
        showgrid=True,
        autorange='reversed'  # Invert X-axis direction (wavenumber decreasing)
    ),
    legend=dict(
        yanchor="top",
        y=1,  # Position legend vertically
        xanchor="left",
        x=-0.25,  # Position legend to the left of the plot
        bordercolor="Black",
        borderwidth=1
    ),
    hovermode="x unified",
    template="plotly_white"
)

# Show the plot
fig.show()
```
